```python
target_imgset = target_imgsetlist[0]
target_imgset.select_images(filter = 'g')
target_img_g = target_imgset.target_images[0]
target_imgset.select_images(filter = 'r')
target_img_r = target_imgset.target_images[0]
```

```python
# Run for all target_images
target_cataloglist_g = []
target_cataloglist_r = []
visualize = False
verbose = False
for target_imgset in target_imgsetlist:
    target_imgset.select_images(filter = 'g')
    target_img_g = target_imgset.target_images[0]
    target_imgset.select_images(filter = 'r')
    target_img_r = target_imgset.target_images[0]
    target_catalog_g = target_img_g.photometry_sex(
        target_bkg = target_img_g.bkgmap,
        target_bkgrms = target_img_g.bkgrms,
        visualize = visualize,
        verbose = verbose)
    
    target_img_g, target_catalog_g, _, _ = target_img_g.photometric_calibration(
        target_catalog_g,
        mag_lower = 12, 
        mag_upper = 17.5, 
        visualize=  visualize,
        verbose = verbose)
    
    target_catalog_r = target_img_r.photometry_sex(
        target_bkg = target_img_r.bkgmap,
        target_bkgrms = target_img_r.bkgrms,
        visualize = visualize,
        verbose = verbose)
    
    target_img_r, target_catalog_r, _, _ = target_img_r.photometric_calibration(
        target_catalog_r,
        mag_lower = 12, 
        mag_upper = 17.5, 
        visualize=  visualize,
        verbose = verbose)

    target_catalog_r.select_sources(target_ra, target_dec)
    target_cataloglist_r.append(target_catalog_r)
    target_catalog_g.select_sources(target_ra, target_dec)
    target_cataloglist_g.append(target_catalog_g)
    
```

```python
from ezphot.methods import PhotometricCalibration
photcal = PhotometricCalibration()
for target_catalog_g, target_catalog_r in zip(target_cataloglist_g, target_cataloglist_r):
    target_cat_g_calibrated = photcal.apply_color_terms(
        target_img_g,
        target_catalog_g,
        target_catalog_r
    )
    target_cat_r_calibrated = photcal.apply_color_terms(
        target_img_r,
        target_catalog_r,
        target_catalog_g
    )
```

```python
%matplotlib inline
from ezphot.dataobjects import LightCurve, CatalogSet
photspec = LightCurve(CatalogSet(target_cataloglist_g))
photspec.plt_params.scatter_markersize = 15
fig, _, ax, tbl = photspec.plot(target_ra, target_dec,
flux_key = 'MAGSKY_APER_2',
fluxerr_key = 'MAGERR_APER_2',
zperr_key = 'ZPERR_APER_2')
fig = fig['g|7DT']
ax = ax['g|7DT']
```

```python
from astropy.time import Time
for target_catalog_g in target_cataloglist_g:
    target_catalog_g.select_sources(target_ra, target_dec)
    target_catalog_g.target_data['ORIGIN_MAGSKY_APER_2']
    obsdate = Time(target_catalog_g.info.obsdate, format=  'isot').datetime
    ax.scatter(obsdate, target_catalog_g.target_data['ORIGIN_MAGSKY_APER_2'], c = 'r')
```
